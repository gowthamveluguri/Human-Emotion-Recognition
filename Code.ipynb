```python
import cv2
import numpy as np

# Load two different images
prev_img = cv2.imread('ima1.jpg')
next_img = cv2.imread('ima3.webp')

# Convert images to grayscale
prev_gray = cv2.cvtColor(prev_img, cv2.COLOR_BGR2GRAY)
next_gray = cv2.cvtColor(next_img, cv2.COLOR_BGR2GRAY)

# Compute optical flow using Lucas-Kanade method
flow = cv2.calcOpticalFlowFarneback(prev_gray, next_gray, None, 0.5, 3, 15, 3, 5, 1.2, 0)

# Convert flow to polar coordinates (magnitude and angle)
mag, ang = cv2.cartToPolar(flow[..., 0], flow[..., 1])

# Create empty image for optical flow visualization
h, w = prev_gray.shape
flow_img = np.zeros((h, w, 3), dtype=np.uint8)

# Set color based on direction
hue = ang * 180 / np.pi / 2
flow_img[..., 0] = hue
flow_img[..., 1] = 255
flow_img[..., 2] = cv2.normalize(mag, None, 0, 255, cv2.NORM_MINMAX)

# Convert HSV to BGR for visualization
flow_img = cv2.cvtColor(flow_img, cv2.COLOR_HSV2BGR)

# Overlay flow on the first image
flow_lines = cv2.addWeighted(prev_img, 1, flow_img, 2, 0)

# Display result
cv2.imshow('Optical Flow', flow_lines)
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import cv2
import numpy as np

# Load two different images
prev_img = cv2.imread('ima1.jpg')
next_img = cv2.imread('ima3.webp')

# Convert images to grayscale
prev_gray = cv2.cvtColor(prev_img, cv2.COLOR_BGR2GRAY)
next_gray = cv2.cvtColor(next_img, cv2.COLOR_BGR2GRAY)

# Compute optical flow using Lucas-Kanade method
flow = cv2.calcOpticalFlowFarneback(prev_gray, next_gray, None, 0.5, 3, 15, 3, 5, 1.2, 0)

# Convert flow to polar coordinates (magnitude and angle)
mag, ang = cv2.cartToPolar(flow[..., 0], flow[..., 1])

# Create empty image for optical flow visualization
h, w = prev_gray.shape
flow_img = np.zeros((h, w, 3), dtype=np.uint8)

# Set color based on direction
hue = ang * 180 / np.pi / 2
flow_img[..., 0] = hue
flow_img[..., 1] = 255
flow_img[..., 2] = cv2.normalize(mag, None, 0, 255, cv2.NORM_MINMAX)

# Convert HSV to BGR for visualization
flow_img = cv2.cvtColor(flow_img, cv2.COLOR_HSV2BGR)

# Overlay flow on the first image
flow_lines = cv2.addWeighted(prev_img, 1, flow_img, 2, 0)

# Resize the output window to a smaller size
resized_flow_lines = cv2.resize(flow_lines, (0, 0), fx=0.5, fy=0.5)

# Display result
cv2.imshow('Optical Flow', resized_flow_lines)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
