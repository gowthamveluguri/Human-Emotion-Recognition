```python
import cv2
import numpy as np

# Load two different images
prev_img = cv2.imread('im2.jpg')
next_img = cv2.imread('im3.jpg')

# Convert images to grayscale
prev_gray = cv2.cvtColor(prev_img, cv2.COLOR_BGR2GRAY)
next_gray = cv2.cvtColor(next_img, cv2.COLOR_BGR2GRAY)

# Compute optical flow using Lucas-Kanade method
flow = cv2.calcOpticalFlowFarneback(prev_gray, next_gray, None, 0.5, 3, 15, 3, 5, 1.2, 0)

# Display optical flow
h, w = prev_gray.shape
flow_img = np.zeros((h, w, 3), dtype=np.uint8)
flow_img[..., 0] = 255
mag, ang = cv2.cartToPolar(flow[..., 0], flow[..., 1])
flow_img[..., 1] = 255
flow_img[..., 2] = cv2.normalize(mag, None, 0, 255, cv2.NORM_MINMAX)
flow_img = cv2.cvtColor(flow_img, cv2.COLOR_HSV2BGR)

# Display result
cv2.imshow('Optical Flow', flow_img)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
